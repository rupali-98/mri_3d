```python
# %pip install -q plotly

import nibabel as nib
import numpy as np
import plotly.graph_objects as go

case = "hippocampus_222"
root = "data/Task04_Hippocampus"
img = nib.as_closest_canonical(nib.load(f"{root}/imagesTr/{case}.nii.gz"))
lbl = nib.as_closest_canonical(nib.load(f"{root}/labelsTr/{case}.nii.gz"))
data = img.get_fdata()
mask = lbl.get_fdata().astype(int)
sx, sy, sz = img.header.get_zooms()[:3]

# physical coordinates (mm) of every voxel
X, Y, Z = np.mgrid[0:data.shape[0], 0:data.shape[1], 0:data.shape[2]]
X, Y, Z = (X * sx).ravel(), (Y * sy).ravel(), (Z * sz).ravel()

fig = go.Figure([
    # whole MRI as a see-through cloud; only brighter tissue is drawn
    go.Volume(x=X, y=Y, z=Z, value=data.ravel(),
              isomin=np.percentile(data, 60), isomax=data.max(),
              opacity=0.07, surface_count=12, colorscale="gray", showscale=False, name="MRI"),
    # hippocampus parts as solid surfaces
    go.Isosurface(x=X, y=Y, z=Z, value=(mask == 1).ravel().astype(float), isomin=0.5, isomax=1,
                  colorscale=[[0, "orangered"], [1, "orangered"]], showscale=False, name="anterior"),
    go.Isosurface(x=X, y=Y, z=Z, value=(mask == 2).ravel().astype(float), isomin=0.5, isomax=1,
                  colorscale=[[0, "steelblue"], [1, "steelblue"]], showscale=False, name="posterior"),
])
fig.update_layout(height=700, title=f"{case} (orange = anterior, blue = posterior)",
                  scene=dict(aspectmode="data", xaxis_title="R (mm)", yaxis_title="A (mm)", zaxis_title="S (mm)"))
fig.show()
```
